```python
import numpy as np
import pandas as pf
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# Ejercicio ejemplo regresión lineal ventas
y = np.array([[7000],[9000],[5000],[11000],[10000],[13000]])
X = np.array([[1],[2],[3],[4],[5],[6]])
```

```python
# Crear modelo vacio
modelo = LinearRegression()
```

```python
modelo.fit(X,y)
```

```python
modelo.coef_[0]
```

```python
modelo.intercept_
```

```python
modelo.predict([[7]])
```

```python
plt.plot(X,y)
```
