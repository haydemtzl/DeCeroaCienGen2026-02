```python
# Importar librerias
import numpy as np
import pandas as pd
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [2,4,1,8,7]

plt.plot(x,y)
```

```python
# Para verificar si hay una relación lineal
correlacion = np.corrcoef(x,y)
correlacion
```

1 - Relación lineal positiva perfecta

0.8 - Muy fuerte

0.5 Moderada

0 no hay relacion lineal

-0.8 Relacion fuerte negativa

-1 Relacion negactiva perfecta


```python
# Ejercicio ejemplo regresión lineal VENTAS
y = np.array([[7000],[9000],[5000],[11000],[10000],[13000]])
X = np.array([[1],[2],[3],[4],[5],[6]])
```

```python
#Crear modelo vacio
regre_ventas = linear_model.LinearRegression()
```

```python
regre_ventas.fit(X,y)
```

```python
# Ver la pendiente
regre_ventas.coef_
```

```python
# Ver la intersección de la linea
regre_ventas.intercept_
```

```python
# Voy a generar predicciones
regre_ventas.predict([[7]])
```

```python
# Graficar 
plt.plot(X,y)
```
