```python
# --- clasificador de calificaciones
def clasificar_calificacion(calificacion):
    if calificacion == 100:
        return "Perfecto"
    elif calificacion >= 90:
        return "Excelente"
    elif calificacion >= 70:
        return "Aprobado"
    else:
        return "Reprobado"
```

```python
clasificar_calificacion(100)
```

```python
clasificar_calificacion(75)
```

```python
clasificar_calificacion(50)
```

```python
# -- Filtrar calificaciones
def obtener_aprobados(calificaciones):
    aprobados = []
    for calificacion in calificaciones:
        if calificacion >= 70:
            aprobados.append(calificacion)
    return aprobados
```

```python
calificaciones = [50,80,95,60,75]
obtener_aprobados(calificaciones)
```

```python
obtener_aprobados([100,86,72,60,98])
```

```python
# -- Filtrar calificaciones reprobatorias
def obtener_reprobados(calificaciones):
    aprobados = []
    for calificacion in calificaciones:
        if calificacion < 70:
            aprobados.append(calificacion)
    return aprobados
```

```python
obtener_reprobados(calificaciones)
```

```python
# -- Filtrar calificaciones con cal aprobatoria
def obtener_aprobados(calificaciones, aprob):
    aprobados = []
    for calificacion in calificaciones:
        if calificacion >= aprob:
            aprobados.append(calificacion)
    return aprobados
```

```python
obtener_aprobados(calificaciones, 80)
```

```python
obtener_aprobados(calificaciones, 50)
```

```python
# -- Total de ventas
# Instrucciones:
# Crear una función que reciba una lista de ventas y devuelva la suma total.

def calcular_total(ventas):
    total = 0
    for venta in ventas:
        total = total + venta
    return total
```

```python
calcular_total([100,200,300])
```

```python
# -- Promedio de ventas
# Instrucciones:
# Crear una función que reciba una lista de ventas y devuelva el promedio.

def promedio(ventas):
    suma = calcular_total(ventas)
    resultado = suma / len(ventas)
    return resultado
```

```python
promedio([100,200,300])
```

```python
# -- Revisar código

def obtener_mayores(numeros):
    mayores = []
    for numero in numeros:
        if numero > 10:
            mayores.append(numero)
    return mayores
```

```python
obtener_mayores([5,10,15,20,25])
```

```python
def obtener_medios(numeros):
    medios = []
    for numero in numeros:
        if numero < 15 and numero > 7:
            medios.append(numero)
    return medios
```

```python
obtener_medios([5,8,11,14,20,25])
```

```python
def ejemplo_or(numeros):
    resultado = []
    for numero in numeros:
        if numero > 20 or numero < 12:
            resultado.append(numero)
    return resultado
```

```python
ejemplo_or([5,8,11,14,20,25])
```

```python
def agenda(nombres, telefonos):
    if len(nombres) != len(telefonos):
        print("Las listas son de diferentes tamaños")
    else:
        dicc = {}
        for indice in range(len(nombres)):
            dicc[nombres[indice]]=telefonos[indice]
        return dicc
```

```python
agenda(["Memo","Irving","Luis"], ["3319876543","123891738372","12830982"])
```
