```python
# Para importar librerías uso "import"
import numpy as np
```

```python
# Con los arreglos de numpy ya puedo usarlos como objetos algebraícos en este caso vector
v + v1
```

```python
# np.ones() me permite crear vectores con 1's con la cantidad que le paso a la función, en este caso le pedí 3 unos
v2 = np.ones(3)
v2
```

```python
# para crear un vector normal, uso np.array([-aqui va una lista-]) 
v3 = np.array([1,2,3])
v3
```

```python
# suma de vectores
x+y
```

```python
# resta de vectores
x-y
```

```python
# multiplicacion de vector por escalar
x*2
```

```python
# multiplicacion de vector por escalar
y*3
```

```python
# calculo del producto punto 
sum(x*y)
```

```python
np.dot(x,y)
```

```python
# Vector ortogonal o perpendicular
v1 = np.array([3,4])
v2 = np.array([4,-3])
```

```python
# Con este .reshape(3,4) le digo que me los acomode en 3 renglones y 4 columnas
A = np.arange(1,13).reshape(3,4)
A
```

```python
B = np.arange(8).reshape(4,2)
B
```
