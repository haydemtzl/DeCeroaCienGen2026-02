```python
# Importar las librerias necesarias
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar mis datos
data = pd.read_csv("2titanic.csv")
```

```python
# Para saber cuantas columnas y renglones tiene mi dataset
data.shape
```

```python
# Cuántos renglones tiene cada columa
data.count()
```

```python
col_names = data.columns.tolist() # Convierte los nombres de las columnas en una lista y las guarda en la variable col_names
col_names
```

```python
for col in col_names:
    print("Valores nulos en", col, ":", data[col].isnull().sum()) # Esta funcion hace una sumatoria de todos los valores nulos por columna
```

```python
# Vamos a ver la estadística descriptiva de mi dataset
data.describe()
```

```python
data.head()
```

```python
def sex_transform(x):
    d = {'male':'M','female':'F'}
    return(d[x])

data["Sex"] = data['Sex'].apply(sex_transform)
```

```python
data.head()
```

```python
# Cuántos y cuáles pasajeros se subieron sin pagar?

data[data.Fare == 0] # Muestrame todos los renglones que cumplan la condicion de que la columna fare sea igual a 0
```

```python
# Tabla cruzada entre supervivencia y sexo
pd.crosstab(data.Survived, data.Sex)
```

```python
# Agrupar los datos por clase y sexo, despues solo mostrar la suma de los que sobrevivieron
pclass_sex_survival_count = data.groupby(['Pclass','Sex'])['Survived'].sum()
pclass_sex_survival_count
```

```python
# Tabla cruzada entre Pclass y sexo
pd.crosstab(data.Pclass, data.Sex)
```

```python
# Gráfica de la columna survived
data.Survived.value_counts().plot(kind = 'bar')
plt.title("Sobrevivieron / cuenta total")
```

```python
# Gráfica de la columna survived
data.Survived.value_counts(normalize = True).plot(kind = 'bar', color = "pink")
plt.title("Sobrevivieron / porcentaje total")
```

```python
# Gráfica de sexo y survived en cuenta total
data.Sex[data.Survived == 1].value_counts().plot(kind = 'barh', color = "purple", alpha = 0.7)
plt.title("sobrevivieron totales - M vs F")
```

```python
# Gráfica de supervivencia por clase
data.Pclass[data.Survived == 1].value_counts().plot(kind = 'bar', color = "green")
plt.title("Sobrevivientes por clase - Totales")
```

```python
# Gráfica de supervivencia por clase en porcentajes - pastel
data.Pclass[data.Survived == 1].value_counts(normalize = True).plot(kind = "pie", autopct = "%1.0f%%")
plt.title("Sobrevivientes por clase - porcentajes - pastel")
```

```python
# Gráfica para ver si hay una relación entre tipo de ticket y edad
fig = plt.figure(figsize = (20,10))
for clase in [1,2,3]:
    data.Age[data.Pclass == clase].plot(kind = "kde")
plt.legend(("1ra Clase","2da Clase","3era Clase"))
```
