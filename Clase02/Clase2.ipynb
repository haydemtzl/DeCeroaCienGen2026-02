```python
# Estructuras de control

# Condicionales "==" igual, "!="diferente, "<" menor que, ">" mayor que, "<=" menor que o igual, ">=" mayor que o igual.

felicidad = True

if felicidad == True:
    print("Toy bien felipe")
    resultado = 2 + 2
    if resultado > 3:
        print(resultado)
    else:
        print("No fue mayor que 3")
else:
    print("Toy bien sad")
```
